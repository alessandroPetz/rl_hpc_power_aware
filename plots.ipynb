```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ============================================================
# FIGURE 2 -- Main comparison for RQ1
# Representative configuration:
# threshold = 400 kW
# battery = 3.2 MWh
# renewables = 400/450 kW where applicable
# Sim 1 uses the version WITH battery cost
# ============================================================
plt.rcParams["pdf.fonttype"] = 42   # TrueType fonts (compatibile con LaTeX)
plt.rcParams["ps.fonttype"]  = 42

plt.rcParams.update({
    "font.size": 24,
    "axes.labelsize": 16,
    "axes.titlesize": 16,
    "xtick.labelsize": 18,
    "ytick.labelsize": 14,
    "legend.fontsize": 14
})

def make_figure2():
    labels = ["Sim 0", "Sim 1", "Sim 2"]

    # Representative configuration
    cost_eur = np.array([
        45901.31,   # Sim 0, threshold 400kW
        47874.24,   # Sim 1, threshold 400kW, battery 3.2MWh, with battery cost
        21390.47    # Sim 2, threshold 400kW, battery 3.2MWh, wind/solar 400/450
    ], dtype=float)

    co2_kg = np.array([
        67248.18,   # Sim 0
        70393.10,   # Sim 1
        24285.34    # Sim 2
    ], dtype=float)

    cost_norm = cost_eur / cost_eur[0]
    co2_norm = co2_kg / co2_kg[0]

    cost_delta_pct = (cost_eur / cost_eur[0] - 1.0) * 100.0
    co2_delta_pct = (co2_kg / co2_kg[0] - 1.0) * 100.0

    x = np.arange(len(labels))

    fig, axes = plt.subplots(1, 2, figsize=(10.5, 4.2))

    # (a) Cost
    axes[0].bar(x, cost_norm, edgecolor="black")
    axes[0].set_xticks(x)
    axes[0].set_xticklabels(labels)
    axes[0].set_ylabel("Normalized total cost")
    axes[0].set_title("(a) Total cost")
    axes[0].axhline(1.0, linestyle="--", linewidth=1)
    axes[0].set_ylim(0, max(cost_norm) * 1.18)

    for i, v in enumerate(cost_norm):
        label = "baseline" if i == 0 else f"{cost_delta_pct[i]:+.1f}%"
        axes[0].text(i, v + 0.03, label, ha="center", va="bottom", fontsize=14)

    # (b) CO2
    axes[1].bar(x, co2_norm, edgecolor="black")
    axes[1].set_xticks(x)
    axes[1].set_xticklabels(labels)
    axes[1].set_ylabel(r"Normalized total CO$_2$")
    axes[1].set_title(r"(b) Total CO$_2$")
    axes[1].axhline(1.0, linestyle="--", linewidth=1)
    axes[1].set_ylim(0, max(co2_norm) * 1.18)

    for i, v in enumerate(co2_norm):
        label = "baseline" if i == 0 else f"{co2_delta_pct[i]:+.1f}%"
        axes[1].text(i, v + 0.03, label, ha="center", va="bottom", fontsize=14)

    plt.tight_layout()
    plt.savefig("figure2_rq1_normalized.png", dpi=300, bbox_inches="tight")
    plt.close()


# ============================================================
# FIGURE 3 -- Impact of stricter grid cap for RQ2
# Representative comparison:
# Sim 2 vs Sim 3
# battery = 3.2 MWh
# renewables = 400/450 kW
# thresholds = 200 kW, 400 kW
# ============================================================

def make_figure3():
    thresholds = ["200 kW", "400 kW"]

    # Sim 2
    sim2_cost = np.array([
        23187.74,   # threshold 200kW, battery 3.2MWh, 400/450
        21390.47    # threshold 400kW, battery 3.2MWh, 400/450
    ], dtype=float)

    sim2_co2 = np.array([
        24285.34,
        24285.34
    ], dtype=float)

    # Sim 3
    sim3_cost = np.array([
        25511.83,
        21606.54
    ], dtype=float)

    sim3_co2 = np.array([
        25383.52,
        24433.00
    ], dtype=float)

    # Generator energy from Sim 3
    sim3_gen_energy_wh = np.array([
        10461649.63,
        1337556.504
    ], dtype=float)
    sim3_gen_energy_kwh = sim3_gen_energy_wh / 1000.0

    cost_overhead_pct = (sim3_cost - sim2_cost) / sim2_cost * 100.0
    co2_overhead_pct = (sim3_co2 - sim2_co2) / sim2_co2 * 100.0

    x = np.arange(len(thresholds))
    fig, axes = plt.subplots(1, 3, figsize=(13, 4.2))

    # (a) cost overhead
    axes[0].bar(x, cost_overhead_pct, edgecolor="black")
    axes[0].set_xticks(x)
    axes[0].set_xticklabels(thresholds)
    axes[0].set_ylabel("Cost overhead [%]")
    axes[0].set_title("(a) Cost overhead")
    axes[0].set_ylim(0, max(cost_overhead_pct) * 1.20)

    for i, v in enumerate(cost_overhead_pct):
        axes[0].text(i, v + max(cost_overhead_pct) * 0.04, f"{v:.1f}%", ha="center", va="bottom", fontsize=14)

    # (b) co2 overhead
    axes[1].bar(x, co2_overhead_pct, edgecolor="black")
    axes[1].set_xticks(x)
    axes[1].set_xticklabels(thresholds)
    axes[1].set_ylabel(r"CO$_2$ overhead [%]")
    axes[1].set_title(r"(b) CO$_2$ overhead")
    axes[1].set_ylim(0, max(co2_overhead_pct) * 1.20)

    for i, v in enumerate(co2_overhead_pct):
        axes[1].text(i, v + max(co2_overhead_pct) * 0.04, f"{v:.1f}%", ha="center", va="bottom", fontsize=14)

    # (c) generator energy
    axes[2].bar(x, sim3_gen_energy_kwh, edgecolor="black")
    axes[2].set_xticks(x)
    axes[2].set_xticklabels(thresholds)
    axes[2].set_ylabel("Generator energy [kWh]")
    axes[2].set_title("(c) Backup usage")
    axes[2].set_ylim(0, max(sim3_gen_energy_kwh) * 1.20)

    for i, v in enumerate(sim3_gen_energy_kwh):
        axes[2].text(i, v + max(sim3_gen_energy_kwh) * 0.04, f"{v:.0f}", ha="center", va="bottom", fontsize=14)

    plt.tight_layout()
    plt.savefig("figure3_rq2_overhead.png", dpi=300, bbox_inches="tight")
    plt.close()


# ============================================================
# FIGURE 4 -- Representative dispatch behavior over time
#
# You need a time-series CSV / DataFrame with at least these columns:
#   time
#   load_w
#   renewables_w
#   grid_w
#   battery_w
#   generator_w
#   soc
#
# Convention suggested:
#   battery_w > 0   means battery discharge to load
#   battery_w < 0   means battery charging
#
# You can adapt column names below if needed.
# ============================================================



# ============================================================
# MAIN
# ============================================================

if __name__ == "__main__":
    make_figure2()
    make_figure3()

    print("Generated: figure2_rq1_normalized.png")
    print("Generated: figure3_rq2_overhead.png")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from utils.renewable_real import RenewableModels
from utils.co2 import CarbonIntensityModels
from utils.battery_model import Battery
from utils.generator_model import Generator
from utils.price_model import PriceModel


# ============================================================
# CONFIGURAZIONE SCELTA PER FIGURE 4
# Sim 3
# threshold = 200 kW
# battery = 3.2 MWh
# renewables = 400/450 kW
# ============================================================

def make_figure4_from_csv(csv_path, start=None, end=None):
    df = pd.read_csv(csv_path)

    # Parse time
    df["time"] = pd.to_datetime(df["time"])

    # Optional time window
    if start is not None:
        df = df[df["time"] >= pd.to_datetime(start)]
    if end is not None:
        df = df[df["time"] <= pd.to_datetime(end)]

    # Required columns
    required_cols = ["time", "load_w", "renewables_w", "grid_w", "battery_w", "generator_w", "soc"]
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in CSV: {missing}")

    # Convert to kW for readability
    load_kw = df["load_w"] / 1000.0
    ren_kw = df["renewables_w"] / 1000.0
    grid_kw = df["grid_w"] / 1000.0
    batt_kw = df["battery_w"] / 1000.0
    gen_kw = df["generator_w"] / 1000.0
    soc = df["soc"]

    fig, axes = plt.subplots(3, 1, figsize=(11, 7), sharex=True)

    # Panel 1: load and renewables
    axes[0].plot(df["time"], load_kw, label="Load")
    axes[0].plot(df["time"], ren_kw, label="Renewables")
    axes[0].set_ylabel("Power [kW]")
    axes[0].set_title("(a) Load and renewable availability")
    axes[0].legend()

    # Panel 2: grid, battery, generator
    axes[1].plot(df["time"], grid_kw, label="Grid")
    axes[1].plot(df["time"], batt_kw, label="Battery")
    axes[1].plot(df["time"], gen_kw, label="Generator")
    axes[1].set_ylabel("Power [kW]")
    axes[1].set_title("(b) Dispatch components")
    axes[1].legend()

    # Panel 3: SOC
    axes[2].plot(df["time"], soc, label="SOC")
    axes[2].set_ylabel("SOC")
    axes[2].set_xlabel("Time")
    axes[2].set_title("(c) Battery state of charge")
    axes[2].legend()

    plt.rcParams["pdf.fonttype"] = 42   # TrueType fonts (compatibile con LaTeX)
    plt.rcParams["ps.fonttype"]  = 42

    plt.rcParams.update({
        "font.size": 24,
        "axes.labelsize": 16,
        "axes.titlesize": 16,
        "xtick.labelsize": 10,
        "ytick.labelsize": 10,
        "legend.fontsize": 14
    })


    plt.tight_layout()
    plt.savefig("figure4_timeline.png", dpi=300, bbox_inches="tight")
    plt.close()


THRESHOLD = 200000
BATTERY_CAPACITY = 6_400_000
P_MAX_REN = (400000, 450000)

OUTPUT_DIR = "outputSim"
TIMESERIES_CSV = os.path.join(OUTPUT_DIR, "sim3_timeseries_200kW_6p4MWh_400_450.csv")
FIGURE_PATH = "figure4_timeline.png"

SOLAR_COST_EUR_PER_KWH = 0.05
WIND_COST_EUR_PER_KWH = 0.04

SOLAR_CO2_G_PER_KWH = 50.0
WIND_CO2_G_PER_KWH = 34.0


# ============================================================
# LOAD DATASET
# ============================================================

df = pd.read_csv("csvs/cluster_power_only_nodes_30days.csv")
df["time"] = pd.to_datetime(df["time"], utc=True)
df["dt_hours"] = df["time"].diff().dt.total_seconds() / 3600
df["dt_hours"] = df["dt_hours"].fillna(0)

price_model = PriceModel()
df["price_base"], df["price_high"] = price_model.prices_from_df(df)

rm = RenewableModels()
df["P_wind"] = rm.wind_from_openmeteo(df, P_MAX_REN[0])
df["P_solar"] = rm.solar_from_openmeteo(df, P_MAX_REN[1])
df["P_ren"] = df["P_solar"] + df["P_wind"]

cm = CarbonIntensityModels(csv_file="csvs/carbon_intensity_IT-NORTH-2020.csv")
df["co2_intensity"] = cm.co2_from_csv(df)


# ============================================================
# BATTERY + GENERATOR
# ============================================================

battery = Battery(
    capacity_wh=BATTERY_CAPACITY,
    initial_charge_wh=BATTERY_CAPACITY / 2,
    max_charge_rate_w=BATTERY_CAPACITY,
    max_discharge_rate_w=BATTERY_CAPACITY,
)

generator = Generator(
    max_power_w=500000,
    min_power_w=50000,
    efficiency=0.4,
    fuel_cost_per_wh=0.00025,
    co2_g_per_kwh=450
)


# ============================================================
# STORAGE FOR RESULTS
# ============================================================

os.makedirs(OUTPUT_DIR, exist_ok=True)

timeseries_rows = []

total_cost = 0.0
total_co2_g = 0.0

sum_E_green_total_Wh = 0.0
sum_green_cost = 0.0
sum_green_co2_g = 0.0

sum_batt_cost = 0.0
sum_batt_co2_g = 0.0

sum_grid_cost = 0.0
sum_grid_co2_g = 0.0

sum_gen_cost = 0.0
sum_gen_co2_g = 0.0

sum_E_from_grid = 0.0
sum_E_from_gen = 0.0
sum_E_from_batt = 0.0
sum_E_batt_charged = 0.0


# ============================================================
# SIMULATION LOOP
# ============================================================

for idx, row in df.iterrows():
    COST_LOW = row["price_base"]
    COST_HIGH = row["price_high"]
    CO2_G_PER_KWH_GRID = row["co2_intensity"]

    P_load = row["power"]       # W
    P_wind = row["P_wind"]      # W
    P_solar = row["P_solar"]    # W
    P_ren = row["P_ren"]        # W
    dt = row["dt_hours"]        # h
    t = row["time"]

    if dt <= 0:
        timeseries_rows.append({
            "time": t,
            "load_w": P_load,
            "renewables_w": P_ren,
            "grid_w": 0.0,
            "battery_w": 0.0,
            "generator_w": 0.0,
            "soc": battery.info()["SOC"],
            "curtailment_wh": 0.0
        })
        continue

    # --------------------------------------------------------
    # 0) GREEN COST / CO2
    # --------------------------------------------------------
    E_wind_total = P_wind * dt
    E_solar_total = P_solar * dt
    E_green_total = E_wind_total + E_solar_total

    green_cost = (
        (E_wind_total / 1000.0) * WIND_COST_EUR_PER_KWH +
        (E_solar_total / 1000.0) * SOLAR_COST_EUR_PER_KWH
    )

    green_co2_g = (
        (E_wind_total / 1000.0) * WIND_CO2_G_PER_KWH +
        (E_solar_total / 1000.0) * SOLAR_CO2_G_PER_KWH
    )

    sum_E_green_total_Wh += E_green_total
    sum_green_cost += green_cost
    sum_green_co2_g += green_co2_g

    # init per-step
    E_charged = 0.0
    E_from_batt = 0.0
    E_curtail = 0.0

    E_from_gen = 0.0
    E_from_grid = 0.0

    cost_grid = 0.0
    cost_gen = 0.0
    co2_grid = 0.0
    co2_gen = 0.0

    P_grid = 0.0
    P_gen = 0.0
    P_batt_plot = 0.0  # >0 scarica, <0 carica

    # --------------------------------------------------------
    # 1) use renewables to cover the load
    # --------------------------------------------------------
    P_from_ren = min(P_load, P_ren)
    P_excess_ren = max(P_ren - P_from_ren, 0.0)

    # --------------------------------------------------------
    # 1a) surplus renewable -> battery
    # --------------------------------------------------------
    if P_excess_ren > 0:
        E_excess_available = P_excess_ren * dt
        E_charged = battery.charge(P_excess_ren, dt)
        E_curtail = max(E_excess_available - E_charged, 0.0)

        # per la figura: batteria in carica = potenza negativa
        P_batt_plot = -(E_charged / dt) if dt > 0 else 0.0

    else:
        # ----------------------------------------------------
        # 2) deficit -> battery
        # ----------------------------------------------------
        P_remaining = P_load - P_from_ren
        E_needed = P_remaining * dt

        E_from_batt = battery.discharge(P_remaining, dt)
        E_needed_after_batt = max(E_needed - E_from_batt, 0.0)

        P_after_batt = E_needed_after_batt / dt if dt > 0 else 0.0

        # per la figura: batteria in scarica = potenza positiva
        P_batt_plot = (E_from_batt / dt) if dt > 0 else 0.0

        # ----------------------------------------------------
        # 3) GRID + GENERATOR
        # ----------------------------------------------------
        use_generator = 1

        P_grid_base = min(P_after_batt, THRESHOLD)
        P_peak = max(P_after_batt - THRESHOLD, 0.0)

        if use_generator:
            P_gen = min(P_peak, generator.max_power_w)
        else:
            P_gen = 0.0

        P_grid_peak = P_peak - P_gen
        P_grid = P_grid_base + P_grid_peak

        E_from_gen = P_gen * dt
        E_from_grid = P_grid * dt

        E_base = P_grid_base * dt
        E_peak = P_grid_peak * dt

        cost_grid = E_base * COST_LOW + E_peak * COST_HIGH
        cost_gen = generator.get_cost(E_from_gen)

        co2_grid = (E_from_grid / 1000.0) * CO2_G_PER_KWH_GRID
        co2_gen = generator.get_co2(E_from_gen / 1000.0)

    # --------------------------------------------------------
    # 4) BATTERY COST / CO2
    # --------------------------------------------------------
    E_batt_throughput = E_charged + E_from_batt

    batt_cost = battery.step_cost(dt, E_batt_throughput)
    batt_co2_g = battery.step_co2_g(dt, E_batt_throughput)

    sum_batt_cost += batt_cost
    sum_batt_co2_g += batt_co2_g

    sum_grid_cost += cost_grid
    sum_grid_co2_g += co2_grid
    sum_gen_cost += cost_gen
    sum_gen_co2_g += co2_gen

    sum_E_from_grid += E_from_grid
    sum_E_from_gen += E_from_gen
    sum_E_from_batt += E_from_batt
    sum_E_batt_charged += E_charged

    # --------------------------------------------------------
    # 5) STEP TOTALS
    # --------------------------------------------------------
    step_cost = green_cost + batt_cost + cost_grid + cost_gen
    step_co2_g = green_co2_g + batt_co2_g + co2_grid + co2_gen

    total_cost += step_cost
    total_co2_g += step_co2_g

    # --------------------------------------------------------
    # save row for figure
    # --------------------------------------------------------
    timeseries_rows.append({
        "time": t,
        "load_w": P_load,
        "renewables_w": P_ren,
        "grid_w": P_grid,
        "battery_w": P_batt_plot,
        "generator_w": P_gen,
        "soc": battery.info()["SOC"],
        "curtailment_wh": E_curtail
    })

    battery.step(dt)


# ============================================================
# SAVE TIMESERIES CSV
# ============================================================

df_ts = pd.DataFrame(timeseries_rows)
df_ts.to_csv(TIMESERIES_CSV, index=False)

print(f"Saved timeseries to: {TIMESERIES_CSV}")


# ============================================================
# MAKE FIGURE 4
# ============================================================

# scegli una finestra temporale più leggibile del mese intero
# cambia queste date se vuoi
START_TIME = df_ts["time"].iloc[0]
#END_TIME = df_ts["time"].iloc[min(len(df_ts) - 1, 24 * 3)]  # finestra iniziale corta se non sai ancora cosa scegliere
END_TIME = df_ts["time"].iloc[-1]

# Se vuoi 48h reali e il timestep e' 20s:
# steps_48h = int((48 * 3600) / 20)
# END_TIME = df_ts["time"].iloc[min(len(df_ts) - 1, steps_48h)]

df_plot = df_ts[(df_ts["time"] >= START_TIME) & (df_ts["time"] <= END_TIME)].copy()

df_plot["time"] = pd.to_datetime(df_plot["time"])
df_plot["load_kw"] = df_plot["load_w"] / 1000.0
df_plot["renewables_kw"] = df_plot["renewables_w"] / 1000.0
df_plot["grid_kw"] = df_plot["grid_w"] / 1000.0
df_plot["battery_kw"] = df_plot["battery_w"] / 1000.0
df_plot["generator_kw"] = df_plot["generator_w"] / 1000.0

fig, axes = plt.subplots(3, 1, figsize=(11, 7), sharex=True)

# (a) load / renewables / threshold
axes[0].plot(df_plot["time"], df_plot["load_kw"], label="Load")
axes[0].plot(df_plot["time"], df_plot["renewables_kw"], label="Renewables")
axes[0].axhline(THRESHOLD / 1000.0, linestyle="--", linewidth=1, label="Grid threshold")
axes[0].set_ylabel("Power [kW]")
axes[0].set_title("(a) Load, renewables, and grid threshold")
axes[0].legend(loc="upper right")

# (b) grid / battery / generator
axes[1].plot(df_plot["time"], df_plot["grid_kw"], label="Grid")
axes[1].plot(df_plot["time"], df_plot["battery_kw"], label="Battery")
axes[1].plot(df_plot["time"], df_plot["generator_kw"], label="Generator")
axes[1].set_ylabel("Power [kW]")
axes[1].set_title("(b) Dispatch components")
axes[1].legend(loc="upper right")

# (c) SOC
axes[2].plot(df_plot["time"], df_plot["soc"], label="SOC")
axes[2].set_ylabel("SOC")
axes[2].set_xlabel("Time")
axes[2].set_title("(c) Battery state of charge")
axes[2].legend(loc="upper right")

plt.rcParams["pdf.fonttype"] = 42   # TrueType fonts (compatibile con LaTeX)
plt.rcParams["ps.fonttype"]  = 42

plt.rcParams.update({
    "font.size": 24,
    "axes.labelsize": 16,
    "axes.titlesize": 16,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 14
})


plt.tight_layout()
plt.savefig(FIGURE_PATH, dpi=300, bbox_inches="tight")
plt.close()

print(f"Saved figure to: {FIGURE_PATH}")
```
